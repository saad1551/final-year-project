```python
import requests

urls = {
    "train.json": "https://huggingface.co/datasets/osunlp/Mind2Web/resolve/main/data/train.json?download=true",
    "validation.json": "https://huggingface.co/datasets/osunlp/Mind2Web/resolve/main/data/validation.json?download=true",
    "test.json": "https://huggingface.co/datasets/osunlp/Mind2Web/resolve/main/data/test.json?download=true"
}

for filename, url in urls.items():
    r = requests.get(url)
    with open(filename, "wb") as f:
        f.write(r.content)
    print(f"Downloaded {filename}")
```
